# CF landscape — what has been solved, what remains

Phase 3 exploratory analysis. This notebook **reads dated outputs** from `python -m cf_atlas landscape`. It does not invent numbers.

Layers used below: **FACT** (CFF published aggregates), **ANALYSIS** (counts from the ClinicalTrials.gov snapshot), **INTERPRETATION** (not a ranking), **HYPOTHESIS** (to test in Phase 4).

Not medical advice. Not a securities recommendation.

```python
from pathlib import Path
import pandas as pd
from cf_atlas.paths import processed_dir, repo_root

land = processed_dir() / "landscape"
claims = pd.read_csv(land / "landscape_claims.csv")
solved = pd.read_csv(land / "solved_unsolved.csv")
needs = pd.read_csv(land / "need_snapshots.csv")
by_year = pd.read_csv(land / "trials_by_year.csv")
ncfb = pd.read_csv(land / "ncfb_flagged_trials.csv")
print("claims", len(claims), "ncfb flagged", len(ncfb))
print("figures", sorted(p.name for p in (repo_root() / "assets" / "figures").glob("*.png")))
```
